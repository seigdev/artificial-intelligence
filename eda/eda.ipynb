## performing EDA

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv('data.csv')
```

```python
# fetching sample 5 rows without replacement
sample = data.sample(n=5, replace=False) 
print(sample.iloc[:, -3:])
```

```python
# pandas df approach
# basic scatter plot
plt.plot(data['column1'], data['column2'], ls="", marker='o')
```

```python
# scatter plot with multiple layers
plt.plot(data['column1'], data['column2'], ls="", marker='o', label='label1')
plt.plot(data['column1'], data['column3'], ls="", marker='o', label='label2')
plt.legend()
```

```python
# histogram
plt.hist(data['column1'], bins=25)
```

```python
# customizing plots
# matplotlib syntax
fig, ax = plt.subplots()
ax.barh(np.arange(10), data['column1'].iloc[:10])

# set position of ticks and ticks labels
ax.set_yticks(np.arange(0.4,10.4,1.0))
ax.set_yticklabels(np.arange(1,11))
ax.set(xlabel='xlabel',ylabel='ylabel', title='title') 
```

```python
# customizing plots by group
data.groupby('column1').mean().plot(color=['red', 'blue','black','green'], fontsize=10.0, figsize=(4,4))
```

```python
# pair plot for feature correlation
sns.pairplot(data, hue='target', size=3)
```

```python
# seaborn hexbin plot
sns.jointplot(x='column1', y='column2', kind='hex', color='blue', size=5)
```

```python
# seaborn plot, facet grid
plot = sns.FacetGrid(data, col='column1', margin_titles=True)
plot.map(plt.hist, 'column2',color='green', bins=20)

# second plot
plot = sns.FacetGrid(data, col='column2 ', margin_titles=True)
plot.map(plt.hist, 'column2',color='blue', bins=20)
```
