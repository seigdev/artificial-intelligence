## Policies for missing data
- removing duplicate data and null values rows entirely
- replacing with substituted values. fill missing data with the most common value.
- create a category for missing values, masking the data.

`Outliers are observations in data that is distant from other observations. it is important to remember that some outliers are informative and provide insights into the data.`

- we can use plots; histogram, density plot, box plot to find this outliers. it gives statistics and residuals of the data.

```python
import seaborn as sns
import numpy as np
```

```python
# plot a histogram and density plot
data = sns.load_dataset("penguins")
sns.displot(data, bins=20)

# plot a boxplot
sns.boxplot(data=data)
```

```python
# detecting outliers: statistical method

# calculate the IQR
q25, q50, q75 = np.percentile(data, [25, 50, 75])
iqr = q75 - q25

# calculate the min and max limits to detect outliers
min = q25 - 1.5 * iqr
max = q75 + 1.5 * iqr

print(f"Min: {min}, {q25}, {q50}, {q75}, Max: {max}")

# identify the points
[x for x in data if x > max] 
```
