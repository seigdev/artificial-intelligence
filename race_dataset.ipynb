```python
import os
import tarfile
import csv
import json
```

```python
# create files directory if it does not exists
if not os.path.isdir("RACE/"):
    tar =  tarfile.open("RACE.tar.gz", "r:gz") 
    tar.extractall()
    tar.close()
else:
    print("Files already extracted!")
```

The article field in each txt document in the RACE dataset is consolidated into a single race_articles.csv file for training purpose.

```python
data_path = "RACE/"

if not os.path.isfile("race_articles.csv"):
    document_file = open("race_articles.csv", mode="w", newline="")
    csv_writer = csv.writer(document_file)

    # Writing the header
    csv_writer.writerow(['articles'])

    for folder in os.listdir(data_path)[::-1]:
        for sub_folder in os.listdir(os.path.join(data_path, folder)):
            for file in os.listdir(os.path.join(data_path, folder, sub_folder)):
                file_path = os.path.join(data_path, folder, sub_folder, file)

                with open(file_path, "r") as json_file:
                    json_data = json.load(json_file)

                # Writing the 'article' in 'articles' column
                csv_writer.writerow([json_data['article']])

    document_file.close()
else:
    print("race_articles.csv already exists!")
```
